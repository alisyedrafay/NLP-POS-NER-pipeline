```python
nlp = spacy.load("en_core_web_sm")
nlp
```

```python
nlp.pipe_names
```

```python
doc = nlp("Tesla Inc  is going to acquire twitter for $45 billion")
doc
```

```python
for ent in doc.ents:
  print(ent.text, "---->", ent.label_, spacy.explain(ent.label_))
```

```python
spacy.explain("MONEY")
```

```python
from spacy import displacy
displacy.render(doc, style="ent")
```

```python
nlp.pipe_names
```

```python
nlp.pipe_labels["ner"]
```

```python
sent = nlp("Elon Musk is a billionaire entrepreneur whose vast wealth is largely driven by his leadership and ownership in Tesla, one of the world’s most valuable electric vehicle companies.")
sent
```
